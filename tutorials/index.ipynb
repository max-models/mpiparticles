# Tutorials

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Moving markers between ranks](/mpiparticles/tutorials/moving-markers/) loads markers into a decomposed box, pushes them and migrates them to their owners.
- [Loading markers](/mpiparticles/tutorials/loading-markers/) draws markers (Maxwellian and flux speeds, a quiet start, importance sampling) and emits them each step from plane and volume sources.
- [Moments and binning](/mpiparticles/tutorials/moments-and-binning/) sums the velocity moments of the markers and bins them in phase space.
- [Pushing markers](/mpiparticles/tutorials/pushing-markers/) pushes markers through electric and magnetic fields with the Boris scheme, in 1D3V, 2D3V and 3D3V.
- [Relativistic markers](/mpiparticles/tutorials/relativistic-markers/) pushes markers close to the speed of light: gyration, acceleration in an electric field and the relativistic E×B drift.
- [Building the docs](/mpiparticles/development/docs/) explains how to add a tutorial and control its output.
