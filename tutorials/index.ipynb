# Tutorials

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Moving markers between ranks](/mpiparticles/tutorials/moving-markers/) loads markers into a decomposed box, pushes them and migrates them to their owners.
- [Moments and binning](/mpiparticles/tutorials/moments-and-binning/) sums the velocity moments of the markers and bins them in phase space.
- [Building the docs](/mpiparticles/development/docs/) explains how to add a tutorial and control its output.
